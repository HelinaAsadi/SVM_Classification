```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
import pandas as pd
```

```python
X, y = make_circles(n_samples=1000, noise=0.15, factor=0.3, random_state=0)

# coverting ndarrays to a dataframe
df = pd.DataFrame(X, columns=["x1" , "x2"])
df["class"] = y
print(df)
```

```python
# plotting
fig, ax = plt.subplots()
ax.scatter(df["x1"], df["x2"] , c= df["class"])
ax.set_title("synthetic dataset")
ax.set_xlabel("x1")
ax.set_ylabel("x2")
plt.show()
```

```python
# inspecting data

print(f"shape : {df.shape}")

print(f"data types: {df.dtypes}")

print(f"class ditribution : {df["class"].value_counts()}")
```
